```python
display(HTML(r'''<iframe title="JavaScript" sandbox="allow-scripts" srcdoc="&lt;!doctype html&gt;&lt;html lang=&quot;es&quot;&gt;&lt;head&gt;&lt;meta charset=&quot;utf-8&quot;&gt;&lt;meta name=&quot;viewport&quot; content=&quot;width=device-width,initial-scale=1&quot;&gt;&lt;style&gt;

*{box-sizing:border-box}body{margin:0;background:#e8d9bb;color:#403229;font:14px/1.4 &quot;Special Elite&quot;,&quot;Courier New&quot;,monospace;background-image:radial-gradient(#6c513518 .65px,transparent .65px);background-size:5px 5px}
button,input,select{font:inherit}button,a,input,select{touch-action:manipulation}button,a{cursor:pointer}button{border:1px solid #573e2b;border-radius:2px;padding:6px 10px;background:#76563a;color:#fff4dd;box-shadow:1px 2px 0 #573e2b;font-weight:400}button:hover{background:#59432f}button:active{box-shadow:none;transform:translate(1px,1px)}button:focus-visible,a:focus-visible,input:focus-visible,select:focus-visible{outline:2px dashed #914a31;outline-offset:3px}a{color:#765039;text-underline-offset:3px}input,select{border:1px solid #ac9673;border-radius:1px;padding:7px;background:#fbf3e1;color:#403229;min-width:0}input::placeholder{color:#8b775f}input[type=checkbox]{accent-color:#76563a;width:15px;height:15px}
.shell{max-width:1140px;margin:0 auto;padding:14px 16px;background:#f3e8d0;border:1px solid #bcaa88;box-shadow:3px 3px 0 #6e57371a}.eyebrow{letter-spacing:.1em;font-size:9px;text-transform:uppercase;color:#79644e}.top{display:flex;justify-content:space-between;align-items:center;gap:12px;padding-bottom:9px;margin-bottom:11px;border-bottom:3px double #8d7758}.top h1{font-size:25px;line-height:1.1;margin:5px 0 3px;font-weight:400}.top .muted{font-size:10px}.muted{color:#7b6853}.pill{display:inline-block;padding:6px 9px;border:1px dashed #927955;background:#e7d5af;font-size:11px}
.toolbar{display:grid;grid-template-columns:minmax(0,1fr) 145px 140px;gap:8px;margin-bottom:6px}.toolbar label{display:grid;gap:3px;font-size:11px}.layout{display:grid;grid-template-columns:minmax(0,1fr) 260px;gap:15px;align-items:start}.grid{display:grid;grid-template-columns:repeat(3,minmax(0,1fr));gap:10px}.card{border:1px solid #b7a181;background:#fbf3e1;overflow:hidden;box-shadow:1px 2px 0 #9b825525}.art{height:91px;background:#e2d3b4;display:flex;align-items:center;justify-content:center;border-bottom:1px solid #c6b28c;background-image:radial-gradient(#8d6b3a14 1px,transparent 1px);background-size:4px 4px}.art img{display:block;width:100%;height:100%;object-fit:contain;padding:2px}.card:nth-child(2n) .art{background-color:#dcd3b8}.card-body{padding:8px 9px}.card h3{font-size:14px;line-height:1.25;margin:4px 0 5px;min-height:35px;font-weight:400;overflow-wrap:anywhere}.card strong{font-size:14px;font-weight:400}.category{font-size:9px;text-transform:uppercase;letter-spacing:.06em;color:#8c674b}.row{display:flex;justify-content:space-between;align-items:center;gap:7px}.actions{display:flex;gap:5px;align-items:center;justify-content:space-between;margin-top:8px}.actions a{font-size:11px}.actions button{font-size:11px;padding:5px 8px}
.cart{background:#efe1c3;border:1px solid #b7a181;border-top:3px solid #84694e;padding:12px}.cart h2{font-size:21px;margin:0 0 8px;font-weight:400}.cart-list{list-style:none;margin:0;padding:0}.cart-item{padding:6px 0;border-bottom:1px dashed #bca781;display:grid;grid-template-columns:minmax(0,1fr) auto;gap:5px;align-items:center}.cart-item .row{display:block;min-width:0}.cart-item strong{display:block;font-size:11px;font-weight:400;overflow-wrap:anywhere}.cart-item .small{display:block;font-size:10px;white-space:nowrap}.qty{display:flex;align-items:center;gap:6px;font-size:11px}.qty button{padding:1px 6px;background:#e0cda8;color:#403229;box-shadow:none}.total{font-size:23px;margin:9px 0}.discount{display:flex;align-items:center;gap:6px;font-size:11px;margin-top:11px}.ghost{background:transparent;color:#6f432b;box-shadow:none;font-size:11px}.ghost:hover{background:#dfc9a5}.small{font-size:11px}.empty{padding:14px;color:#7b6853;border:1px dashed #b7a181;grid-column:1/-1}
.detail{background:#fbf3e1;border:1px solid #b7a181;padding:17px}.detail .art{height:180px;margin:12px 0;border:1px solid #c6b28c}.detail h2{font-size:25px;font-weight:400;line-height:1.3;margin:10px 0}.detail p{font-size:13px}.detail .row{flex-wrap:wrap}.status{font-size:11px;margin:7px 0;color:#7b6853}footer{margin:11px 0 0;padding-top:7px;border-top:3px double #8d7758;font-size:9px;color:#7b6853;letter-spacing:.03em}
@media(max-width:740px){.shell{padding:8px;border:0}.top{margin-bottom:8px;padding-bottom:7px}.top h1{font-size:22px}.top .eyebrow{font-size:8px}.top .muted{font-size:9px}.toolbar{grid-template-columns:minmax(0,1fr) 105px 100px;gap:5px}.toolbar label{font-size:10px}input,select{padding:6px;font-size:11px}.layout{grid-template-columns:1fr;gap:9px}.grid{gap:6px}.art{height:44px}.card-body{padding:5px}.card h3{font-size:12px;min-height:30px;margin:3px 0 2px}.card strong{font-size:12px}.category{font-size:8px;letter-spacing:0}.actions{gap:4px;margin-top:4px}.actions a{font-size:10px}.actions button{font-size:10px;padding:4px 5px}.cart{padding:8px 10px;display:grid;grid-template-columns:minmax(0,1fr) auto;gap:5px 8px}.cart h2{font-size:18px;margin:0;grid-column:1;grid-row:1}.cart&gt;.total{grid-column:2;grid-row:1;margin:0;font-size:19px}.cart&gt;div:first-of-type{grid-column:1/-1;grid-row:2}.cart-list{display:grid;grid-template-columns:repeat(3,minmax(0,1fr));gap:5px}.cart-item{display:flex;flex-direction:column;align-items:flex-start;padding:3px 0;gap:2px}.cart-item strong{font-size:9px;line-height:1.2}.cart-item .small{font-size:9px}.qty{gap:5px;font-size:10px}.qty button{padding:0 5px;font-size:10px}.cart .discount{grid-column:1;grid-row:3;margin:0;font-size:10px}.cart .ghost{grid-column:2;grid-row:3;font-size:10px;padding:4px 6px}.cart p{margin:0;font-size:10px}footer{margin-top:8px;font-size:8px}.detail{padding:12px}.detail .art{height:140px}}
@media(max-width:350px){.toolbar{grid-template-columns:minmax(0,1fr) 95px 90px}.shell{padding:8px}.actions{flex-wrap:wrap}}

&lt;/style&gt;&lt;/head&gt;&lt;body&gt;

&lt;main class=&quot;shell&quot;&gt;
&lt;header class=&quot;top&quot;&gt;&lt;div&gt;&lt;div class=&quot;eyebrow&quot;&gt;EQUIPO DE SUPERVIVENCIA / 01&lt;/div&gt;&lt;h1&gt;REFUGIO 84&lt;/h1&gt;&lt;span class=&quot;muted&quot;&gt;JavaScript&lt;/span&gt;&lt;/div&gt;&lt;span class=&quot;pill&quot; id=&quot;header-count&quot; aria-live=&quot;polite&quot;&gt;0 productos&lt;/span&gt;&lt;/header&gt;
&lt;div class=&quot;layout&quot;&gt;&lt;section&gt;&lt;div id=&quot;catalog-view&quot;&gt;&lt;div class=&quot;toolbar&quot;&gt;&lt;label&gt;Buscar&lt;input id=&quot;query&quot; placeholder=&quot;Pip-Boy, agua, radio…&quot; type=&quot;search&quot;&gt;&lt;/label&gt;&lt;label&gt;Categoría&lt;select id=&quot;category&quot; aria-label=&quot;Categoría&quot;&gt;&lt;option value=&quot;Todas&quot;&gt;Todas&lt;/option&gt;&lt;option&gt;Electrónica&lt;/option&gt;&lt;option&gt;Protección&lt;/option&gt;&lt;option&gt;Suministros&lt;/option&gt;&lt;/select&gt;&lt;/label&gt;&lt;label&gt;Orden&lt;select id=&quot;sort&quot; aria-label=&quot;Orden&quot;&gt;&lt;option value=&quot;name&quot;&gt;Nombre&lt;/option&gt;&lt;option value=&quot;price&quot;&gt;Menor precio&lt;/option&gt;&lt;/select&gt;&lt;/label&gt;&lt;/div&gt;&lt;p class=&quot;status&quot; id=&quot;results&quot; aria-live=&quot;polite&quot;&gt;&lt;/p&gt;&lt;div id=&quot;catalog&quot; class=&quot;grid&quot;&gt;&lt;/div&gt;&lt;/div&gt;&lt;div id=&quot;detail-view&quot; hidden&gt;&lt;/div&gt;&lt;/section&gt;
&lt;aside class=&quot;cart&quot;&gt;&lt;h2&gt;Tu carrito&lt;/h2&gt;&lt;div id=&quot;cart&quot; aria-live=&quot;polite&quot;&gt;&lt;/div&gt;&lt;label class=&quot;discount&quot;&gt;&lt;input id=&quot;discount&quot; type=&quot;checkbox&quot;&gt; Aplicar 10 % de descuento&lt;/label&gt;&lt;div class=&quot;total&quot; id=&quot;total&quot; aria-live=&quot;polite&quot;&gt;&lt;/div&gt;&lt;button id=&quot;clear&quot; class=&quot;ghost&quot;&gt;Vaciar carrito&lt;/button&gt;&lt;/aside&gt;&lt;/div&gt;
&lt;footer&gt;REFUGIO 84 / Inventario de demostración / USD&lt;/footer&gt;&lt;/main&gt;

&lt;script&gt;

// Todas las descargas y conversiones ocurren en JavaScript,
// dentro del mismo iframe que contiene la tienda.
const URL_BASE = &quot;https://raw.githubusercontent.com/artemioadaysolvers/-JavaScript-vs-TypeScript/main/assets/&quot;;
const URL_FUENTE = &quot;https://raw.githubusercontent.com/google/fonts/main/apache/specialelite/SpecialElite-Regular.ttf&quot;;
const nombres_imagenes = [
  &quot;pip-boy.png&quot;, &quot;mascara-antirradiacion.png&quot;, &quot;botella-agua.png&quot;,
  &quot;botiquin-medico.png&quot;, &quot;radio-portatil.png&quot;, &quot;linterna-campo.png&quot;
];

async function descargar_asset(url) {
  const response = await fetch(url);
  if (!response.ok) throw new Error(&#x27;Error HTTP &#x27; + response.status + &#x27;: &#x27; + url);
  return new Uint8Array(await response.arrayBuffer());
}

function convertir_base64(data) {
  let binary = &#x27;&#x27;;
  for (let i = 0; i &lt; data.length; i += 8192) {
    binary += String.fromCharCode(...data.subarray(i, i + 8192));
  }
  return btoa(binary);
}

function detectar_mime(data) {
  if (data[0] === 0x89 &amp;&amp; data[1] === 0x50 &amp;&amp; data[2] === 0x4e &amp;&amp; data[3] === 0x47)
    return &#x27;image/png&#x27;;
  if (data[0] === 0xff &amp;&amp; data[1] === 0xd8 &amp;&amp; data[2] === 0xff)
    return &#x27;image/jpeg&#x27;;
  if (new TextDecoder().decode(data.subarray(0, 1000)).trimStart().includes(&#x27;&lt;svg&#x27;))
    return &#x27;image/svg+xml&#x27;;
  return &#x27;image/png&#x27;;
}

window.REFUGIO_ASSETS_READY = (async () =&gt; {
  // Crear el mismo payload, sin usar Python ni compartir datos entre iframes.
  const imagenes_base64 = {};
  for (const name of nombres_imagenes) {
    const data = await descargar_asset(URL_BASE + name);
    imagenes_base64[name] = {
      mime: detectar_mime(data),
      content: convertir_base64(data)
    };
  }
  let fuente_base64 = null;
  if (URL_FUENTE) {
    const font_data = await descargar_asset(URL_FUENTE);
    fuente_base64 = {content: convertir_base64(font_data)};
  }
  const payload = {images: imagenes_base64, font: fuente_base64};
  window._refugio_payload = payload;

  // Mantener intacta la aplicación y asignar sus imágenes originales.
  const bytes = encoded =&gt; Uint8Array.from(atob(encoded), char =&gt; char.charCodeAt(0));
  const urls = Object.fromEntries(Object.entries(payload.images || {}).map(([name, image]) =&gt;
    [name, URL.createObjectURL(new Blob([bytes(image.content)], {type:image.mime}))]));
  function applyImages() {
    for (const image of document.querySelectorAll(&#x27;img[data-asset]&#x27;)) {
      const url = urls[image.dataset.asset];
      if (url &amp;&amp; image.getAttribute(&#x27;src&#x27;) !== url) image.setAttribute(&#x27;src&#x27;,url);
    }
  }
  applyImages();
  const observer = new MutationObserver(applyImages);
  observer.observe(document.documentElement, {childList:true,subtree:true,attributes:true,attributeFilter:[&#x27;data-asset&#x27;]});
  if (payload.font &amp;&amp; payload.font.content) {
    const face = new FontFace(&#x27;Special Elite&#x27;, bytes(payload.font.content).buffer);
    await face.load();
    document.fonts.add(face);
  }
  await new Promise(resolve =&gt; requestAnimationFrame(() =&gt; requestAnimationFrame(resolve)));
  applyImages();
  window.addEventListener(&#x27;beforeunload&#x27;, () =&gt; {
    observer.disconnect();
    Object.values(urls).forEach(url =&gt; URL.revokeObjectURL(url));
  }, {once:true});
  console.log(&#x27;✓ Payload listo en window._refugio_payload&#x27;);
  return urls;
})().catch(error =&gt; { console.error(&#x27;Imágenes de Refugio 84:&#x27;, error); throw error; });

&lt;/script&gt;&lt;script&gt;


const products = [{&quot;id&quot;: 1, &quot;name&quot;: &quot;pip-boy&quot;, &quot;category&quot;: &quot;Electrónica&quot;, &quot;price&quot;: 149, &quot;image&quot;: &quot;pip-boy.png&quot;, &quot;description&quot;: &quot;Ordenador de muñeca ficticio con pantalla verde y carcasa de estética retro militar.&quot;}, {&quot;id&quot;: 2, &quot;name&quot;: &quot;Máscara antirradiación&quot;, &quot;category&quot;: &quot;Protección&quot;, &quot;price&quot;: 39, &quot;image&quot;: &quot;mascara-antirradiacion.png&quot;, &quot;description&quot;: &quot;Máscara ficticia de estética postapocalíptica para completar el equipo del refugio.&quot;}, {&quot;id&quot;: 3, &quot;name&quot;: &quot;Botella de agua&quot;, &quot;category&quot;: &quot;Suministros&quot;, &quot;price&quot;: 12, &quot;image&quot;: &quot;botella-agua.png&quot;, &quot;description&quot;: &quot;Cantimplora de estilo militar ochentero, con tapón de rosca y funda verde oliva.&quot;}, {&quot;id&quot;: 4, &quot;name&quot;: &quot;Botiquín médico&quot;, &quot;category&quot;: &quot;Suministros&quot;, &quot;price&quot;: 35, &quot;image&quot;: &quot;botiquin-medico.png&quot;, &quot;description&quot;: &quot;Estuche de primeros auxilios de estética retro, con cruz roja y cierre metálico.&quot;}, {&quot;id&quot;: 5, &quot;name&quot;: &quot;Radio portátil&quot;, &quot;category&quot;: &quot;Electrónica&quot;, &quot;price&quot;: 49, &quot;image&quot;: &quot;radio-portatil.png&quot;, &quot;description&quot;: &quot;Radio de campo de inspiración ochentera, con antena telescópica y dial analógico.&quot;}, {&quot;id&quot;: 6, &quot;name&quot;: &quot;Linterna de campo&quot;, &quot;category&quot;: &quot;Electrónica&quot;, &quot;price&quot;: 29, &quot;image&quot;: &quot;linterna-campo.png&quot;, &quot;description&quot;: &quot;Linterna clásica de carcasa verde oliva y reflector ancho, inspirada en el equipo de los años 80.&quot;}];
const money = (n) =&gt; new Intl.NumberFormat(&#x27;es&#x27;, {style:&#x27;currency&#x27;,currency:&#x27;USD&#x27;}).format(n);
const cart = {};
let discount = false;
const el = (id) =&gt; {
  const node = document.getElementById(id);
  if (!node) throw new Error(&#x27;No existe el elemento: &#x27; + id);
  return node;
};
const input = (id) =&gt; el(id);
const select = (id) =&gt; el(id);
const escapeHTML = (s) =&gt; s.replace(/[&amp;&lt;&gt;&quot;&#x27;]/g, c =&gt; ({&#x27;&amp;&#x27;:&#x27;&amp;amp;&#x27;,&#x27;&lt;&#x27;:&#x27;&amp;lt;&#x27;,&#x27;&gt;&#x27;:&#x27;&amp;gt;&#x27;,&#x27;&quot;&#x27;:&#x27;&amp;quot;&#x27;,&quot;&#x27;&quot;:&#x27;&amp;#39;&#x27;})[c]);

function renderCatalog() {
  const query = input(&#x27;query&#x27;).value.toLocaleLowerCase();
  const category = select(&#x27;category&#x27;).value;
  const visible = products.filter(p =&gt; p.name.toLocaleLowerCase().includes(query) &amp;&amp; (category === &#x27;Todas&#x27; || p.category === category));
  visible.sort((a,b) =&gt; select(&#x27;sort&#x27;).value === &#x27;price&#x27; ? a.price-b.price : a.name.localeCompare(b.name, &#x27;es&#x27;));
  el(&#x27;results&#x27;).textContent = visible.length + &#x27; productos encontrados&#x27;;
  el(&#x27;catalog&#x27;).innerHTML = visible.length ? visible.map(p =&gt; `
    &lt;article class=&quot;card&quot; data-product=&quot;${p.id}&quot;&gt;&lt;div class=&quot;art&quot; aria-hidden=&quot;true&quot;&gt;&lt;img data-asset=&quot;${p.image}&quot; alt=&quot;&quot;&gt;&lt;/div&gt;&lt;div class=&quot;card-body&quot;&gt;
    &lt;div class=&quot;category&quot;&gt;${escapeHTML(p.category)}&lt;/div&gt;&lt;h3&gt;${escapeHTML(p.name)}&lt;/h3&gt;&lt;strong&gt;${money(p.price)}&lt;/strong&gt;
    &lt;div class=&quot;actions&quot;&gt;&lt;a href=&quot;#producto/${p.id}&quot;&gt;Detalle&lt;/a&gt;&lt;button data-add=&quot;${p.id}&quot; aria-label=&quot;Agregar ${escapeHTML(p.name)}&quot;&gt;Agregar&lt;/button&gt;&lt;/div&gt;&lt;/div&gt;&lt;/article&gt;
  `).join(&#x27;&#x27;) : &#x27;&lt;p class=&quot;empty&quot;&gt;No hay productos con esos filtros.&lt;/p&gt;&#x27;;
}

function renderCart() {
  const items = products.filter(p =&gt; (cart[p.id] || 0) &gt; 0);
  const count = items.reduce((n,p) =&gt; n + cart[p.id], 0);
  const subtotal = items.reduce((n,p) =&gt; n + p.price*cart[p.id], 0);
  el(&#x27;header-count&#x27;).textContent = count + &#x27; productos&#x27;;
  el(&#x27;total&#x27;).textContent = money(subtotal * (discount ? .9 : 1));
  el(&#x27;cart&#x27;).innerHTML = items.length ? &#x27;&lt;ul class=&quot;cart-list&quot;&gt;&#x27; + items.map(p =&gt; `
    &lt;li class=&quot;cart-item&quot;&gt;&lt;div class=&quot;row&quot;&gt;&lt;strong&gt;${escapeHTML(p.name)}&lt;/strong&gt;&lt;span class=&quot;small&quot;&gt;${money(p.price*cart[p.id])}&lt;/span&gt;&lt;/div&gt;
    &lt;div class=&quot;qty&quot;&gt;&lt;button data-minus=&quot;${p.id}&quot; aria-label=&quot;Quitar una unidad de ${escapeHTML(p.name)}&quot;&gt;−&lt;/button&gt;&lt;span&gt;${cart[p.id]}&lt;/span&gt;&lt;button data-add=&quot;${p.id}&quot; aria-label=&quot;Agregar una unidad de ${escapeHTML(p.name)}&quot;&gt;+&lt;/button&gt;&lt;/div&gt;&lt;/li&gt;
  `).join(&#x27;&#x27;) + &#x27;&lt;/ul&gt;&#x27; : &#x27;&lt;p class=&quot;muted small&quot;&gt;Tu carrito está vacío.&lt;/p&gt;&#x27;;
}

function changeQuantity(id, delta) {
  cart[id] = Math.max(0, (cart[id] || 0) + delta);
  renderCart();
}

function route() {
  const match = location.hash.match(/^#producto\/(\d+)$/);
  const product = match ? products.find(p =&gt; p.id === Number(match[1])) : undefined;
  el(&#x27;catalog-view&#x27;).hidden = Boolean(product);
  el(&#x27;detail-view&#x27;).hidden = !product;
  if (product) el(&#x27;detail-view&#x27;).innerHTML = `
    &lt;section class=&quot;detail&quot;&gt;&lt;a href=&quot;#&quot;&gt;← Volver al catálogo&lt;/a&gt;&lt;div class=&quot;art&quot; aria-hidden=&quot;true&quot;&gt;&lt;img data-asset=&quot;${product.image}&quot; alt=&quot;&quot;&gt;&lt;/div&gt;
    &lt;div class=&quot;eyebrow&quot;&gt;${escapeHTML(product.category)}&lt;/div&gt;&lt;h2&gt;${escapeHTML(product.name)}&lt;/h2&gt;&lt;p&gt;${escapeHTML(product.description)}&lt;/p&gt;
    &lt;div class=&quot;row&quot;&gt;&lt;strong&gt;${money(product.price)}&lt;/strong&gt;&lt;button data-add=&quot;${product.id}&quot;&gt;Agregar al carrito&lt;/button&gt;&lt;/div&gt;&lt;p class=&quot;status&quot;&gt;Equipo de estética retro para el inventario del refugio.&lt;/p&gt;&lt;/section&gt;`;
}

document.addEventListener(&#x27;click&#x27;, event =&gt; {
  if (!(event.target instanceof Element)) return;
  const link = event.target.closest(&#x27;a&#x27;);
  if (link &amp;&amp; (link.getAttribute(&#x27;href&#x27;) || &#x27;&#x27;).startsWith(&#x27;#&#x27;)) {
    event.preventDefault();
    location.hash = (link.getAttribute(&#x27;href&#x27;) || &#x27;&#x27;).slice(1);
    return;
  }
  const button = event.target.closest(&#x27;button&#x27;);
  if (!button) return;
  if (button.dataset.add) changeQuantity(Number(button.dataset.add), 1);
  if (button.dataset.minus) changeQuantity(Number(button.dataset.minus), -1);
});
input(&#x27;query&#x27;).addEventListener(&#x27;input&#x27;, renderCatalog);
select(&#x27;category&#x27;).addEventListener(&#x27;change&#x27;, renderCatalog);
select(&#x27;sort&#x27;).addEventListener(&#x27;change&#x27;, renderCatalog);
input(&#x27;discount&#x27;).addEventListener(&#x27;change&#x27;, () =&gt; { discount = input(&#x27;discount&#x27;).checked; renderCart(); });
el(&#x27;clear&#x27;).addEventListener(&#x27;click&#x27;, () =&gt; { Object.keys(cart).forEach(id =&gt; delete cart[Number(id)]); renderCart(); });
window.addEventListener(&#x27;hashchange&#x27;, route);
renderCatalog(); renderCart(); route();

&lt;/script&gt;&lt;/body&gt;&lt;/html&gt;" style="width:100%;height:720px;border:0"></iframe>'''))
```
