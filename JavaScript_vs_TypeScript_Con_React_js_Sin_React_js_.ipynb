```python
# CELDA INICIAL — Versión limpia y directa
import base64, builtins, urllib.request

URL_BASE = "https://raw.githubusercontent.com/artemioadaysolvers/-JavaScript-vs-TypeScript/main/assets/"

URL_FUENTE = "https://raw.githubusercontent.com/google/fonts/main/apache/specialelite/SpecialElite-Regular.ttf"

nombres_imagenes = [
    "pip-boy.png",
    "mascara-antirradiacion.png",
    "botella-agua.png",
    "botiquin-medico.png",
    "radio-portatil.png",
    "linterna-campo.png"
]

def descargar_asset(url):
    req = urllib.request.Request(url, headers={'User-Agent': 'Refugio84-Colab/1.0'})
    with urllib.request.urlopen(req) as response:
        return response.read()

# Cargar imágenes a base64

imagenes_base64 = {}

for name in nombres_imagenes:
    data = descargar_asset(URL_BASE + name)

    # Detección básica de tipo MIME
    if data.startswith(b'\x89PNG\r\n\x1a\n'): mime = 'image/png'
    elif data.startswith(b'\xff\xd8\xff'): mime = 'image/jpeg'
    elif b'<svg' in data.lstrip()[:1000]: mime = 'image/svg+xml'
    else: mime = 'image/png'

    imagenes_base64[name] = {
        'mime': mime,
        'content': base64.b64encode(data).decode('ascii')
    }

# Cargar fuente

fuente_base64 = None

if URL_FUENTE:
    font_data = descargar_asset(URL_FUENTE)
    fuente_base64 = {
        'content': base64.b64encode(font_data).decode('ascii')
    }

# Asignar payload global que leerá tu iframe

builtins._refugio_payload = {
    'images': imagenes_base64,
    'font': fuente_base64
}

print('✓ Payload listo en builtins._refugio_payload')
```
